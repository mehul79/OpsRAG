# OpsRAG: embed the corpus on a GPU
Chunks the docs and embeds them with `bge-small-en-v1.5`, then gives you `chunks.jsonl` + `vectors.npy` to load into Qdrant locally with `python backend/load.py`.

**Before running:** Runtime → Change runtime type → **T4 GPU**.

## 1. Upload the bundle
Upload `colab/opsrag_colab.zip` from the project folder (about the size of the docs, a few tens of MB).

```
from google.colab import files
up = files.upload()  # pick opsrag_colab.zip
```
